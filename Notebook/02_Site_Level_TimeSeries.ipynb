```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("MIG_clean.csv", parse_dates=["date"])
df = df.set_index("date")
```

```python
df.head()
```

Using the most performing site to check the first Site level time series. 

```python
site = "SITE_025"
df_site = df[df["site_id"] == site].copy()
```

```python
df_site = df_site.sort_values("date")
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.figure(figsize=(16,6), facecolor="none")

plt.plot(df_site.index, df_site['consumed_tonnes'], color="#0A1A44", linewidth=2)

plt.title(f"{site} — Daily Cement Consumption - Site 025", fontsize=24, fontweight="bold")
plt.xlabel("Date", fontsize=18, fontweight="bold")
plt.ylabel("Tonnes Consumed", fontsize=18, fontweight="bold")

# Format the x-axis dates properly
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=3))   # show every 3 months
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))   # format as YYYY-MM

plt.xticks(rotation=45, fontsize=16)
plt.yticks(fontsize=16)

plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
```
